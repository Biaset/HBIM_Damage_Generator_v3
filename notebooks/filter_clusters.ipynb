```python
filtered_cluster = enriched_cluster[enriched_cluster["point_count"] > 100]
color_map = {
    'Lichen': (0.4549, 0.4549, 0.0157), #2.0 
    'Plant': (0, 0.666667, 0.0), #4.0
    'Moss': (0, 1, 0), #3.0
    'Bacteria': (0, 0, 0) #1.0
}

filtered_cluster[['Rf', 'Gf', 'Bf']] = filtered_cluster['damage_type'].map(color_map).apply(pd.Series)
filtered_cluster
```

```python
filtered_cluster.to_csv(r"data/output/filtered_cluster_attributes.csv", index=False)
```
